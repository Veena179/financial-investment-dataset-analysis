```python
import pandas as pd
df = pd.read_csv("/Data_set 2 - Copy.csv")
duration_map = {
    "Less than 1 year": 0.5,
    "1-3 years": 2,
    "3-5 years": 4,
    "More than 5 years": 6
}
df['Duration_numeric'] = df['Duration'].map(duration_map)
average_duration = df['Duration_numeric'].mean()
average_duration
```

```python
df['Duration'].value_counts().plot(kind='bar')
import matplotlib.pyplot as plt
plt.title("Investment Duration Distribution")
plt.xlabel("Duration")
plt.ylabel("Number of Participants")
plt.show()
```
